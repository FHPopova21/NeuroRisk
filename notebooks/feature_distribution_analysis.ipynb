# Feature Distribution Analysis (C) - Box Plots

## Objective
Visually inspect the separation power of features using Box Plots. 
We expect a good feature to have distinct, non-overlapping boxes ("whiskers" can overlap slightly) for different classes.

## Classes
- **Healthy** (0)
- **Inter-ictal** (1) (Epileptic patient, no seizure)
- **Seizure** (2) (Epileptic patient, seizure active)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# Set plot style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (8, 6)
```

## 1. Load Data & Decode Labels

```python
data_path = '../data/processed/Bonn_EEG_Train.csv'

df = pd.read_csv(data_path)

# Decode One-Hot to String Labels
y_cols = [c for c in df.columns if c.startswith('y_')]
class_map = {0: 'Healthy', 1: 'Inter-ictal', 2: 'Seizure'}

df['Class_Index'] = df[y_cols].idxmax(axis=1).apply(lambda s: int(s.split('_')[1]))
df['Condition'] = df['Class_Index'].map(class_map)

df[['Condition', 'Class_Index']].head()
```

## 2. Box Plot: RMS (Amplitude)
RMS is expected to be much higher for Seizures.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Condition', y='RMS', data=df, palette='Set2')
plt.title('RMS Distribution vs Condition')
plt.show()
```

## 3. Box Plot: Hjorth Mobility (Frequency)
This was our #1 Feature in Part B. Let's see how well it separates.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='Condition', y='Hjorth_Mobility', data=df, palette='Set2')
plt.title('Hjorth Mobility Distribution vs Condition')
plt.show()
```

## 4. Multi-Feature Comparison

```python
features_to_plot = ['RMS', 'Hjorth_Mobility', 'Hjorth_Complexity', 'Hjorth_Activity']
features_to_plot = [f for f in features_to_plot if f in df.columns]

fig, axes = plt.subplots(1, len(features_to_plot), figsize=(20, 5))

for i, feature in enumerate(features_to_plot):
    sns.boxplot(x='Condition', y=feature, data=df, ax=axes[i], palette='Set2')
    axes[i].set_title(feature)

plt.tight_layout()
plt.show()
```
